Confere que o ambiente funciona e dá um primeiro olhar na matriz v0 (5 empresas x 5 anos x 6 DAs).

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import matplotlib.pyplot as plt
from src.data import load_matriz, DAS

df = load_matriz()
df.head()
```

## Maturidade média por empresa e ano

```python
pivot = df.pivot(index='ano', columns='empresa', values='maturidade_media')
pivot.round(2)
```

```python
pivot.plot(marker='o', figsize=(8, 4))
plt.ylabel('Maturidade média (0 a 1)')
plt.title('Maturidade média por empresa, 2021 a 2025')
plt.grid(alpha=.3)
plt.show()
```
